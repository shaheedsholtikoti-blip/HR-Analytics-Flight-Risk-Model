```python
import pandas as pd
import numpy as np
from sklearn.ensemble import RandomForestClassifier
import ipywidgets as widgets
from IPython.display import display, HTML, clear_output

# ==========================================
# 1. ENTERPRISE DATA MODELING & TRAINING
# ==========================================
np.random.seed(42)
n_employees = 2000

# Base DataFrame with Performance Rating added (1-4 scale)
data = pd.DataFrame({
    'Age': np.random.randint(22, 60, n_employees),
    'MonthlyIncome': np.random.randint(40000, 400000, n_employees), 
    'JobSatisfaction': np.random.randint(1, 5, n_employees),
    'WorkLifeBalance': np.random.randint(1, 5, n_employees),
    'OverTime': np.random.choice(['Yes', 'No'], n_employees, p=[0.25, 0.75]),
    'CommuteDistance': np.random.randint(1, 50, n_employees),       
    'YearsAtCompany': np.random.randint(0, 25, n_employees),
    'YearsSinceLastPromotion': np.random.randint(0, 15, n_employees),
    'PerformanceRating': np.random.choice([1, 2, 3, 4], n_employees, p=[0.1, 0.6, 0.2, 0.1])
})
data['YearsSinceLastPromotion'] = data[['YearsAtCompany', 'YearsSinceLastPromotion']].min(axis=1)

# Strategic Risk Weighting
risk_score = (
    np.where(data['OverTime'] == 'Yes', 28, 0) + 
    np.where(data['JobSatisfaction'] <= 2, 22, 0) + 
    np.where(data['WorkLifeBalance'] <= 2, 18, 0) + 
    np.where(data['MonthlyIncome'] < 75000, 15, 0) +
    np.where(data['CommuteDistance'] > 20, 12, 0) +
    np.where(data['YearsSinceLastPromotion'] >= 4, 15, 0) +
    np.where((data['PerformanceRating'] == 4) & (data['YearsSinceLastPromotion'] >= 3), 25, 0) - 
    np.where(data['Age'] > 45, 10, 0) 
)
data['Attrition'] = np.where(risk_score > 65, 1, 0)
data['OverTime_Encoded'] = data['OverTime'].apply(lambda x: 1 if x == 'Yes' else 0)

features = ['Age', 'MonthlyIncome', 'JobSatisfaction', 'WorkLifeBalance', 
            'OverTime_Encoded', 'CommuteDistance', 'YearsAtCompany', 'YearsSinceLastPromotion', 'PerformanceRating']

model = RandomForestClassifier(n_estimators=150, max_depth=7, random_state=42).fit(data[features], data['Attrition'])

# ==========================================
# 2. ENTERPRISE UI & FINANCIAL ENGINE
# ==========================================
display(HTML("""
<style>
    @import url('https://fonts.googleapis.com/css2?family=Helvetica+Neue:wght@300;400;500;700&display=swap');
    
    .jupyter-widgets { font-family: 'Helvetica Neue', Helvetica, Arial, sans-serif !important; }
    .widget-label { font-weight: 500 !important; color: #333333 !important; font-size: 12px !important; text-transform: uppercase; letter-spacing: 0.5px; }
    
    .btn-enterprise { 
        background: #002855 !important; 
        color: #ffffff !important;
        font-size: 13px !important; 
        font-weight: 600 !important; 
        border-radius: 2px !important; 
        border: none !important;
        letter-spacing: 1px !important;
        text-transform: uppercase !important;
        transition: background 0.2s ease !important;
    }
    .btn-enterprise:hover { background: #004080 !important; }
    .fade-in { animation: fadeIn 0.3s ease-in forwards; }
    @keyframes fadeIn { from { opacity: 0; } to { opacity: 1; } }
</style>
"""))

# Corporate Header
header = widgets.HTML("""
<div style="border-bottom: 2px solid #002855; padding-bottom: 15px; margin-bottom: 25px; max-width: 900px; display: flex; justify-content: space-between; align-items: flex-end;">
    <div>
        <div style="font-size: 10px; color: #666; letter-spacing: 1px; margin-bottom: 5px;">CONFIDENTIAL // HR STRATEGY DASHBOARD</div>
        <h2 style="margin: 0; font-family: 'Helvetica Neue', sans-serif; font-weight: 700; font-size: 22px; color: #002855; letter-spacing: -0.5px;">Employee Flight Risk Terminal</h2>
    </div>
    <div style="font-size: 11px; color: #666; font-family: monospace;">MODEL ID: RF-8493-V3 | CURRENCY: INR</div>
</div>
""")

style = {'description_width': '180px'}
layout = widgets.Layout(width='400px', margin='4px 10px')

# Inputs 
age_input = widgets.BoundedIntText(value=32, min=18, max=100, description='Employee Age', style=style, layout=layout)
income_input = widgets.BoundedIntText(value=120000, min=20000, max=1000000, step=10000, description='Monthly Income (₹)', style=style, layout=layout)
distance_input = widgets.BoundedIntText(value=22, min=1, max=150, description='Commute (km)', style=style, layout=layout)
overtime_input = widgets.Dropdown(options=['No', 'Yes'], value='No', description='Frequent Overtime', style=style, layout=layout)
perf_input = widgets.Dropdown(options=[('1 - Needs Improvement', 1), ('2 - Meets Expectations', 2), ('3 - Exceeds', 3), ('4 - Outstanding', 4)], value=2, description='Performance Rating', style=style, layout=layout)

years_co_input = widgets.BoundedIntText(value=5, min=0, max=50, description='Years at Company', style=style, layout=layout)
years_prom_input = widgets.BoundedIntText(value=3, min=0, max=50, description='Yrs since Promotion', style=style, layout=layout)
satisfaction_input = widgets.Dropdown(options=[('1 - Low', 1), ('2 - Medium', 2), ('3 - High', 3), ('4 - Very High', 4)], value=3, description='Job Satisfaction', style=style, layout=layout)
worklife_input = widgets.Dropdown(options=[('1 - Poor', 1), ('2 - Fair', 2), ('3 - Good', 3), ('4 - Excellent', 4)], value=3, description='Work-Life Balance', style=style, layout=layout)

input_grid = widgets.HBox([
    widgets.VBox([age_input, income_input, distance_input, overtime_input, perf_input]),
    widgets.VBox([years_co_input, years_prom_input, satisfaction_input, worklife_input])
], layout=widgets.Layout(margin='0 0 25px 0'))

analyze_btn = widgets.Button(description='Calculate Risk & Financial Loss', layout=widgets.Layout(width='350px', height='40px'))
analyze_btn.add_class('btn-enterprise')

out = widgets.Output()

def on_analyze_clicked(b):
    with out:
        clear_output(wait=True)
        
        ot_val = 1 if overtime_input.value == 'Yes' else 0
        if years_prom_input.value > years_co_input.value:
            display(HTML("<div style='color:#d93025; font-size:12px; font-weight:600;'>ERROR: Years since promotion cannot be higher than total years at company.</div>"))
            return

        input_df = pd.DataFrame(
            [[age_input.value, income_input.value, satisfaction_input.value, worklife_input.value, 
              ot_val, distance_input.value, years_co_input.value, years_prom_input.value, perf_input.value]],
            columns=features
        )
        
        # 1. Predictive Engine
        risk_prob = model.predict_proba(input_df)[0][1] * 100
        
        # 2. Performance-Weighted Financial Impact Engine
        annual_salary = income_input.value * 12
        base_multiplier = 1.5 if income_input.value < 150000 else 2.0
        perf_premium = 0.5 if perf_input.value == 4 else (0.2 if perf_input.value == 3 else 0.0)
        
        replacement_multiplier = base_multiplier + perf_premium
        replacement_cost = annual_salary * replacement_multiplier
        expected_loss = replacement_cost * (risk_prob / 100) 

        # 3. Dynamic Strategy Matrix - Fully restored to cover all variables
        drivers = []
        if perf_input.value == 4 and years_prom_input.value >= 3: 
            drivers.append("<tr><td style='padding:6px 0; border-bottom:1px solid #eee;'>Risk of Losing Top Performer</td><td style='text-align:right; color:#d93025; font-weight:600; border-bottom:1px solid #eee;'>Not promoted in 3+ years</td></tr>")
        elif years_prom_input.value >= 4:
            drivers.append("<tr><td style='padding:6px 0; border-bottom:1px solid #eee;'>Role Stagnation</td><td style='text-align:right; color:#d93025; font-weight:600; border-bottom:1px solid #eee;'>No promotion in 4+ years</td></tr>")
        
        if ot_val == 1: 
            drivers.append("<tr><td style='padding:6px 0; border-bottom:1px solid #eee;'>Heavy Workload</td><td style='text-align:right; color:#d93025; font-weight:600; border-bottom:1px solid #eee;'>Consistent Overtime</td></tr>")
        if satisfaction_input.value <= 2: 
            drivers.append("<tr><td style='padding:6px 0; border-bottom:1px solid #eee;'>Unhappy at Work</td><td style='text-align:right; color:#d93025; font-weight:600; border-bottom:1px solid #eee;'>Low Job Satisfaction</td></tr>")
        if worklife_input.value <= 2: 
            drivers.append("<tr><td style='padding:6px 0; border-bottom:1px solid #eee;'>Burnout Risk</td><td style='text-align:right; color:#d93025; font-weight:600; border-bottom:1px solid #eee;'>Poor Work-Life Balance</td></tr>")
        if distance_input.value > 20: 
            drivers.append("<tr><td style='padding:6px 0; border-bottom:1px solid #eee;'>Commute Fatigue</td><td style='text-align:right; color:#d93025; font-weight:600; border-bottom:1px solid #eee;'>Travels > 20km daily</td></tr>")
        if income_input.value < 75000:
            drivers.append("<tr><td style='padding:6px 0; border-bottom:1px solid #eee;'>Compensation Risk</td><td style='text-align:right; color:#d93025; font-weight:600; border-bottom:1px solid #eee;'>Below standard pay scale</td></tr>")
        
        driver_html = f"<table style='width:100%; font-size:12px; border-collapse:collapse;'>{''.join(drivers)}</table>" if drivers else "<div style='font-size:12px; color:#666;'>No major risks found.</div>"

        # Plain English Action Plans
        if risk_prob > 60 and perf_input.value >= 3:
            status_color, status_text = "#d93025", "CRITICAL: TOP TALENT LEAVING"
            action_plan = "1. Schedule an urgent meeting with the employee today.<br>2. Ensure their salary is highly competitive for the market.<br>3. Fast-track a promotion or give them a major project to lead."
        elif risk_prob > 60:
            status_color, status_text = "#d93025", "HIGH RISK"
            action_plan = "1. Manager needs to intervene this week.<br>2. Discuss their workload and see if they need support."
        elif risk_prob > 30:
            status_color, status_text = "#f29900", "MODERATE RISK"
            action_plan = "1. Check in with them over the next few weeks.<br>2. See if they might be happier switching to a different team."
        else:
            status_color, status_text = "#188038", "SAFE / STABLE"
            action_plan = "1. No urgent action needed. Keep up with regular check-ins."

        # Enterprise Result Dashboard
        result_card = f"""
        <div class="fade-in" style="border: 1px solid #d5d5d5; background: #fafafa; max-width: 900px; font-family: 'Helvetica Neue', sans-serif;">
            
            <!-- Top Metric Bar -->
            <div style="display: flex; background: #ffffff; border-bottom: 1px solid #d5d5d5;">
                <div style="flex: 1; padding: 25px; border-right: 1px solid #d5d5d5;">
                    <div style="font-size: 11px; color: #666; text-transform: uppercase; letter-spacing: 1px; margin-bottom: 8px;">Chance of Leaving</div>
                    <div style="font-size: 42px; font-weight: 300; color: {status_color}; line-height: 1;">{risk_prob:.1f}%</div>
                    <div style="font-size: 12px; font-weight: 700; color: {status_color}; margin-top: 5px; letter-spacing: 0.5px;">STATUS: {status_text}</div>
                </div>
                <div style="flex: 1; padding: 25px;">
                    <div style="font-size: 11px; color: #666; text-transform: uppercase; letter-spacing: 1px; margin-bottom: 8px;">Estimated Financial Cost if they leave</div>
                    <div style="font-size: 42px; font-weight: 300; color: #333; line-height: 1;">₹{expected_loss:,.0f}</div>
                    <div style="font-size: 11px; color: #666; margin-top: 5px;">It costs about {replacement_multiplier}x their salary to replace them.</div>
                </div>
            </div>

            <!-- Bottom Strategy Area -->
            <div style="display: flex;">
                <div style="flex: 1; padding: 25px; background: #ffffff; border-right: 1px solid #d5d5d5;">
                    <div style="font-size: 12px; font-weight: 700; color: #333; text-transform: uppercase; margin-bottom: 15px; border-bottom: 2px solid #002855; padding-bottom: 5px; display:inline-block;">Why they might leave</div>
                    {driver_html}
                </div>
                <div style="flex: 1.5; padding: 25px; background: #f8f9fc;">
                    <div style="font-size: 12px; font-weight: 700; color: #002855; text-transform: uppercase; margin-bottom: 15px; border-bottom: 2px solid #002855; padding-bottom: 5px; display:inline-block;">Recommended HR Action</div>
                    <p style="font-size: 13px; color: #333; line-height: 1.6; margin: 0;">
                        {action_plan}
                    </p>
                </div>
            </div>
            
        </div>
        """
        display(HTML(result_card))

analyze_btn.on_click(on_analyze_clicked)

# Render
display(widgets.VBox([header, input_grid, widgets.HBox([analyze_btn]), out], layout=widgets.Layout(width='100%')))
```
